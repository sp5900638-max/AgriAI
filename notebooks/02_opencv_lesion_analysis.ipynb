# CropSense AI: OpenCV Lesion Quantification & Quality Assurance
Covers:
- Blur & luminance quality checks
- CLAHE illumination equalization (LAB L-channel)
- HSV segmentation of leaf foliage
- Contour extraction, lesion area %, chlorosis ratio, and severity scoring

```python
import cv2
import glob
import matplotlib.pyplot as plt
from src.opencv_utils import (
    check_image_quality,
    preprocess_image,
    enhance_illumination_clahe,
    segment_leaf,
    analyze_lesions
)

images = glob.glob('../data/raw/images/*.png')
sample_path = images[0] if images else None
print("Sample Image:", sample_path)
img = cv2.imread(sample_path)
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
```

## 1. Quality Assessment

```python
quality = check_image_quality(img)
print("Quality Metrics:", quality)
```

## 2. CLAHE & HSV Segmentation

```python
clahe = enhance_illumination_clahe(img)
seg, mask = segment_leaf(clahe)
lesion_data = analyze_lesions(img, mask)

fig, axs = plt.subplots(1, 4, figsize=(16, 4))
axs[0].imshow(img_rgb); axs[0].set_title("Original")
axs[1].imshow(cv2.cvtColor(seg, cv2.COLOR_BGR2RGB)); axs[1].set_title("Segmented Leaf")
axs[2].imshow(mask, cmap='gray'); axs[2].set_title("Foliage Mask")
axs[3].imshow(cv2.cvtColor(lesion_data['contour_overlay'], cv2.COLOR_BGR2RGB)); axs[3].set_title("Lesion Contours")
for ax in axs: ax.axis('off')
plt.tight_layout()
plt.show()

print("Lesion Area %:", lesion_data['lesion_percent'])
print("Chlorosis Ratio:", lesion_data['chlorosis_ratio'])
print("Computed Severity:", lesion_data['severity_level'])
```
