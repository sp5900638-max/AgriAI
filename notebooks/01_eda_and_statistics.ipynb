# CropSense AI: Exploratory Data Analysis & Statistical Modeling
### SDG 2: Zero Hunger
This notebook performs EDA, data cleaning, statistical hypothesis testing (Chi-Square & ANOVA), and unsupervised pattern discovery (PCA + K-Means).

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from src.train_tabular import run_statistical_analysis, discover_patterns

df = pd.read_csv('../data/raw/tabular/crop_health_records.csv')
print(f"Dataset shape: {df.shape}")
df.head()
```

## 1. Class & Seasonal Distributions

```python
plt.figure(figsize=(10, 4))
sns.countplot(data=df, x='disease_label', hue='season', palette='Set2')
plt.title('Disease / Deficiency Occurrence by Season')
plt.xticks(rotation=20)
plt.tight_layout()
plt.show()
```

## 2. Statistical Hypothesis Testing
- **Chi-Square Test**: Condition vs Season
- **One-Way ANOVA**: Environmental variables across risk tiers

```python
stats_results = run_statistical_analysis(df)
print("Chi-square test (Disease vs Season):", stats_results['chi_square'])
print("\nANOVA Results (Top environmental differentiators):")
for var, res in stats_results['anova'].items():
    print(f"{var:<15} F-stat={res['f_stat']:<8} p-value={res['p_value']:.4e} significant={res['significant']}")
```

## 3. Unsupervised Pattern Discovery (PCA + K-Means)

```python
patterns = discover_patterns(df, n_clusters=3)
coords = np.array(patterns['pca_coords'])
labels = patterns['cluster_labels']

plt.figure(figsize=(8, 6))
plt.scatter(coords[:, 0], coords[:, 1], c=labels, cmap='viridis', alpha=0.6)
plt.title('Discovered Environmental Risk Clusters (2D PCA)')
plt.xlabel('Principal Component 1')
plt.ylabel('Principal Component 2')
plt.colorbar(label='Cluster')
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```
